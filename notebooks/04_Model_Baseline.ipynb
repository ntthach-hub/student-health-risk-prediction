#### Baseline Model - CatBoost

In this notebook, we train the first baseline model using CatBoostClassifier.

Workflow:
- Load processed dataset
- Train CatBoost
- Evaluate using Cross Validation
- Predict test data
- Generate submission.csv

### 2. import libraries

```python
import pandas as pd
import numpy as np

from catboost import CatBoostClassifier

from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score

from sklearn.metrics import accuracy_score

import matplotlib.pyplot as plt
```

### 3. load processed data

```python
train = pd.read_csv("../data/processed/processed_train.csv")
test = pd.read_csv("../data/processed/processed_test.csv")
```

### 4 prepare X and y

```python
X = train.drop("health_condition", axis=1)

y = train["health_condition"]

X_test = test.copy()

print(X.shape)
print(y.shape)
print(X_test.shape)
```

### 5. Build baseline model

```python
model = CatBoostClassifier(
    random_state=42,
    verbose=0
)
```

### 6 Cross Validation

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

print(scores)
print("Mean Accuracy:", scores.mean())
```
